```python
from pyspark.sql.types import StringType, StructField, ArrayType, IntegerType, DoubleType, StructType

item_schema = StructType([
    StructField("ItemId", IntegerType(), True),
    StructField("Name", StringType(), True),
    StructField("Price", DoubleType(), True),
    StructField("Quantity", DoubleType(), True),
    StructField("Sum", DoubleType(), True),
    StructField("ProductType", StringType(), True),
])

receipt_schema = StructType([
    StructField("Id", IntegerType(), True),
    StructField("User", IntegerType(), True),
    StructField("OrganizationId", IntegerType(), True),
    StructField("Organization", StringType(), True),
    StructField("LegalForm", StringType(), True),
    StructField("InvoiceId", IntegerType(), True),
    StructField("InvoiceType", IntegerType(), True),
    StructField("InvoiceSum", DoubleType(), True),
    StructField("PaymentType", StringType(), True),
    StructField("CheckSum", DoubleType(), True),
    StructField("CreatedAt", StringType(), True),
    StructField("Items", ArrayType(item_schema), True),
])

print(receipt_schema)
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("GNIVC_Receipts_Analytics") \
    .master("local[*]") \
    .config("spark.jars.packages", "org.apache.spark:spark-sql-kafka-0-10_2.12:3.5.0,org.postgresql:postgresql:42.6.0") \
    .getOrCreate()

print("Spark Version:", spark.version)
```

```python
from pyspark.sql.functions import from_json, col

kafka_df = (
    spark.readStream
    .format("kafka")
    .option("kafka.bootstrap.servers", "localhost:9092")
    .option("subscribe", "receipts-topic")
    .option("startingOffsets", "latest")
    .option("failOnDataLoss", "false")
    .load()
)
receipts_df = kafka_df.select(
    from_json(col("value").cast("string"), receipt_schema).alias("data")
).select("data.*")

receipts_df.printSchema()


```

```python
from pyspark.sql.functions import window, col, sum, count, avg, to_timestamp, approx_count_distinct

# Функция для записи каждого микро-батча в PostgreSQL
def write_to_postgres(batch_df, batch_id):
    batch_df.write \
        .format("jdbc") \
        .option("url", "jdbc:postgresql://localhost:5432/gnivc") \
        .option("dbtable", "basic_metrics") \
        .option("user", "postgres") \
        .option("password", "postgres") \
        .option("driver", "org.postgresql.Driver") \
        .mode("append") \
        .save()
    print(f"Batch {batch_id} written to PostgreSQL")

# Создаем агрегированные данные
receipts_with_ts = receipts_df.withColumn(
    "timestamp",
    to_timestamp(col("CreatedAt"))
)

metrics = receipts_with_ts.groupBy(
    window(col("timestamp"), "1 day")
).agg(
    sum("CheckSum").alias("total_revenue"),
    count("Id").alias("receipts_count"),
    avg("CheckSum").alias("avg_check"),
    approx_count_distinct("User").alias("unique_clients")
).select(
    col("window.start").alias("window_start"),
    col("window.end").alias("window_end"),
    col("total_revenue"),
    col("receipts_count"),
    col("avg_check"),
    col("unique_clients")
)

# Запускаем потоковую обработку с записью в БД через foreachBatch
query = metrics.writeStream \
    .foreachBatch(write_to_postgres) \
    .outputMode("update") \
    .option("checkpointLocation", "/tmp/spark_checkpoint_pg") \
    .start()

print("Streaming started! Data will be written to PostgreSQL.")
query.awaitTermination()
```
